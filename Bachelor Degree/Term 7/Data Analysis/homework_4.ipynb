```python
import matplotlib
import matplotlib.pyplot as plt
import scipy.stats as stats
import pandas as pd
import numpy as np

# Specifying path to the file containing our dataset
DATASET_PATH = 'heart.csv'

# Loading the dataframe
df = pd.read_csv(DATASET_PATH, delimiter=',')
df.head(6) # Showing first 6 entries
```

```python
df.info()
```

## Step 2. Draw some conclusions

### Step 2.1. 
Drop columns we are not interested in

```python
df = df.drop(['restecg', 'exang', 'cp', 'oldpeak', 'slope', 'ca', 'thal'], axis=1)
df = df.rename(columns={
    "trestbps": "blood_pressure", 
    "chol": "cholesterol",
    "fbs": "diabetes",
    "thalach": "heart_rate",
    "target": "heart_disease"
})
df
```

### Step 2.2.
Let us build the correlation matrix!

```python
df.corr()
```

Conclusions we can make:
1. There is a high correlation between the age and the blood pressure/cholesterol level, which is quite reasonable.
2. There is also a pretty high correlation between blood pressure and cholesterol and diabetes.
3. Finally, the presence of heart disease is highly correlated with the heart rate and sex
4. On the other hand, the heart rate and blood pressure has almost no correlation at all!

# Line Plot and Pie Chart

Below are updated plots from **September 3, 2026**.

```python
# --- 1a. Line Plot: Average Maximum Heart Rate by Age ---
plt.figure(figsize=(10, 5))
df.groupby('age')['heart_rate'].mean().plot(kind='line', marker='o', color='teal', linewidth=2)
plt.title('Average Max Heart Rate by Age')
plt.xlabel('Age')
plt.ylabel('Average Heart Rate (bpm)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# --- 1b. Pie Chart: Proportion of Heart Disease Patients ---
plt.figure(figsize=(7, 7))
target_counts = df['heart_disease'].value_counts()
plt.pie(
    target_counts, 
    labels=['No Heart Disease', 'Heart Disease'], 
    colors=['#6baed6', '#fb6a4a'],
    autopct='%1.1f%%', 
    startangle=90,
    explode=(0.05, 0)
)
plt.title('Distribution of Heart Disease Cases')
plt.show()
```

# Bar Chart

```python
# --- 2. Bar Chart: Average Blood Pressure by Sex ---
plt.figure(figsize=(8, 5))
mean_bp = df.groupby('sex')['blood_pressure'].mean()

# Mapping 0 and 1 to Female/Male labels
mean_bp.index = ['Female', 'Male'] 
mean_bp.plot(kind='bar', color=['mediumpurple', 'skyblue'], edgecolor='black', width=0.5)

plt.title('Average Resting Blood Pressure by Sex')
plt.xlabel('Sex')
plt.ylabel('Average Blood Pressure (mm Hg)')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

# Box Plot

```python
# --- 3. Box Plot: Cholesterol Distribution by Heart Disease Status ---
plt.figure(figsize=(8, 6))
df.boxplot(
    column='cholesterol', 
    by='heart_disease', 
    grid=True, 
    patch_artist=True,
    boxprops=dict(facecolor='lightgreen', color='green'),
    medianprops=dict(color='red', linewidth=2)
)

plt.suptitle('')
plt.title('Cholesterol Level Distribution by Heart Disease Presence')
plt.xlabel('Heart Disease (0 = No, 1 = Yes)')
plt.ylabel('Cholesterol (mg/dl)')
plt.show()
```
